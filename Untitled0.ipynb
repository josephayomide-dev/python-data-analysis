```python

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Iris dataset
from sklearn.datasets import load_iris

iris = load_iris()

# Puting data into a table
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = [
    iris.target_names[i]
    for i in iris.target
]

# Display the first five rows
df.head()
```

```python

# 1. How many rows and columns are there?
print("Dataset shape:", df.shape)

# 2. How many flowers of each species?
print("\nFlowers per species:")
print(df["species"].value_counts())

# 3. Summary statistics
print("\nSummary statistics:")
print(df.describe().round(2))
```

```python

# Compare petal measurements across flower species

plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="species",
    s=80
)

plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.tight_layout()
plt.show()
```

```python

# Compare average petal measurements by species

petal_summary = df.groupby("species")[
    ["petal length (cm)", "petal width (cm)"]
].mean().round(2)

print("Average petal measurements by species:")
print(petal_summary)
```
